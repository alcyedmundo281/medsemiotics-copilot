## 2. 📖 Guía Clínica Ampliada: Colitis Ulcerosa (Semana 11)

### 🎯 Resultados de Aprendizaje:
1. **Reconocer el patrón clínico** que hace sospechar colitis ulcerosa (hematoquecia, diarrea sanguinolenta, urgencia, tenesmo).
2. **Excluir diagnósticos infecciosos** antes de atribuir los síntomas a EII (*Clostridioides difficile*, parásitos y bacterias invasivas).
3. **Diferenciar Extensión Anatómica (Montreal)** de **Actividad Clínica (Truelove & Witts)**.
4. **Identificar Colitis Ulcerosa Aguda Grave** y priorizar la hospitalización y el manejo multidisciplinario.
5. **Comparar Colitis Ulcerosa vs. Enfermedad de Crohn** sin basarse en un único hallazgo.

---

### 🗺️ Extensión Anatómica — Clasificación de Montreal

| Categoría | Extensión Anatómica | Clave Docente y Semiótica |
| :---: | :--- | :--- |
| **E1** | **Proctitis Ulcerosa** (limitada al recto) | Predomina sangrado rectal, urgencia y tenesmo; puede cursar con heces formes. |
| **E2** | **Colitis Izquierda** (hasta el ángulo esplénico) | Suma diarrea sanguinolenta franca y dolor abdominal tipo cólico (predominio FII). |
| **E3** | **Colitis Extensa / Pancolitis** (proximal al ángulo esplénico) | Afectación difusa del colon; mayor superficie inflamada y riesgo de complicaciones sistémicas. |

> ⚠️ **Regla de Oro:** La extensión describe **anatomía**, NO **gravedad**. Una proctitis (E1) puede ser sumamente sintomática y una colitis extensa (E3) puede estar en remisión.

---

### ⚖️ Actividad Clínica — Criterios de Truelove y Witts

La **Colitis Ulcerosa Aguda Grave (CUAG)** en adultos se define por:
$$\mathbf{\ge 6\ \text{deposiciones sanguinolentas/día}} \quad + \quad \mathbf{\ge 1\ \text{signo de toxicidad sistémica:}}$$

1. **Temperatura:** $> 37.8\,^\circ\text{C}$
2. **Frecuencia Cardíaca:** $> 90\,\text{lpm}$
3. **Hemoglobina:** $< 10.5\,\text{g/dL}$
4. **Reactantes de Fase Aguda:** $\text{VSG} > 30\,\text{mm/h}$ o $\text{PCR elevada}$

---

### 🔬 Comparación Semiótica: Colitis Ulcerosa vs. Enfermedad de Crohn

| Dimensión | Colitis Ulcerosa | Enfermedad de Crohn |
| :--- | :--- | :--- |
| **Distribución** | Continua, ascendente desde el recto, limitada al colon | Segmentaria / Lesiones salteadas (*skip lesions*), boca al ano |
| **Profundidad** | Mucosa y submucosa superficial | Transmural (riesgo de fístulas, abscesos, estenosis) |
| **Pistas Clínicas** | Hematoquecia, urgencia, pujo y tenesmo | Dolor en FID, diarrea no siempre sanguinolenta, enf. perianal, masa palpable |
| **Urgencia de Alarma** | Colitis aguda grave, megacolon tóxico | Absceso intraabdominal, obstrucción, fístula enterocutánea/vesical |

```python
# Evaluador de Truelove y Witts
def evaluar_truelove_witts(deposiciones_dia, con_sangre=True, temperatura=36.5, fc=75, hb=13.5, vsg=15, pcr=3.0):
    """Calcula y categoriza la actividad clinica segun Truelove y Witts."""
    signos_sistemicos = []
    if temperatura > 37.8:
        signos_sistemicos.append(f"Fiebre ({temperatura} °C > 37.8 °C)")
    if fc > 90:
        signos_sistemicos.append(f"Taquicardia ({fc} lpm > 90 lpm)")
    if hb < 10.5:
        signos_sistemicos.append(f"Anemia ({hb} g/dL < 10.5 g/dL)")
    if vsg > 30 or pcr > 30:
        signos_sistemicos.append(f"Inflamación sistémica (VSG: {vsg} mm/h, PCR: {pcr} mg/L)")
        
    es_grave = (deposiciones_dia >= 6) and con_sangre and (len(signos_sistemicos) >= 1)
    
    html = "<div style='padding:15px; border-radius:8px; margin-top:10px; font-family:sans-serif;"
    if es_grave:
        html += "background-color:#fee2e2; border:2px solid #ef4444; color:#991b1b;'>"
        html += "<h3>🚨 COLITIS ULCEROSA AGUDA GRAVE (CUAG) / URGENCIA HOSPITALARIA</h3>"
        html += f"<p><b>Criterio de deposiciones:</b> {deposiciones_dia} deposiciones/día con sangre (Cumple criterio ≥ 6).</p>"
        html += "<p><b>Signos de toxicidad sistémica identificados:</b></p><ul>"
        for s in signos_sistemicos:
            html += f"<li>{s}</li>"
        html += "</ul>"
        html += "<p><b>Conducta Inmediata:</b> Hospitalización urgente, descartar <i>C. difficile</i> y megacolon tóxico, sigmoidoscopia flexible temprana (no colonoscopia total), monitorización multidisciplinaria.</p>"
    elif deposiciones_dia >= 4:
        html += "background-color:#fef3c7; border:2px solid #f59e0b; color:#92400e;'>"
        html += "<h3>⚠️ ACTIVIDAD MODERADA</h3>"
        html += f"<p>Presenta {deposiciones_dia} deposiciones/día sin toxicidad sistémica grave concomitante.</p>"
    else:
        html += "background-color:#dcfce7; border:2px solid #22c55e; color:#166534;'>"
        html += "<h3>✅ ACTIVIDAD LEVE / REMISIÓN</h3>"
        html += f"<p>Menos de 4 deposiciones/día, sin repercusión sistémica.</p>"
    html += "</div>"
    
    display(HTML(html))

print("Probando evaluador con parámetros del Caso Central HCAM:")
evaluar_truelove_witts(deposiciones_dia=7, con_sangre=True, temperatura=38.0, fc=104, hb=10.2, vsg=35, pcr=45.0)
```

## 5. ⚡ Tres Viñetas Relámpago (Entrenamiento Rápido de Patrones)

Analice rápidamente las tres situaciones anatómicas de la clasificación de Montreal:

1. **Viñeta 1:** Paciente con sangrado rojo rutilante, urgencia y tenesmo; heces formes. Endoscopia muestra mucosa eritematosa y friable estrictamente en los primeros 12 cm desde el margen anal.  
   👉 **Diagnóstico:** **E1 — Proctitis Ulcerosa.**

2. **Viñeta 2:** Paciente con diarrea sanguinolenta de 5 semanas y dolor en flanco izquierdo. Endoscopia revela afectación continua y confluente desde recto hasta el ángulo esplénico; colon transverso y derecho normales.  
   👉 **Diagnóstico:** **E2 — Colitis Izquierda.**

3. **Viñeta 3:** Paciente con pancolitis confirmada que se extiende proximal al ángulo esplénico hasta el ciego.  
   👉 **Diagnóstico:** **E3 — Colitis Extensa / Pancolitis.**

> 💡 **Pregunta Trampa del Docente:** *“¿Cuál de estas tres viñetas corresponde automáticamente a una urgencia grave?”*  
> **Respuesta Clave:** **Ninguna.** La extensión (E1, E2, E3) define anatomía. La gravedad depende de la actividad clínica y de los criterios de Truelove & Witts.
